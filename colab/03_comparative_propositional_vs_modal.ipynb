# ?? Comparative Evaluation: First-Order Propositional vs. Modal Logic Reasoning
[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com)

This notebook compares the internal mechanistic reasoning pathways between **Standard Propositional Logic** (Hong et al. 2025, Chen et al. 2026) and **Modal Logic with Kripke Semantics** (this project).

---
### ?? Key Dimensions Compared:
1. **Circuit Complexity & Head Specialization**:
   - Propositional: 4 core head families (`QRLH`, `QRMH`, `FPH`, `DH`).
   - Modal Logic: 7 head families with modal specialists (`MOH`, `MPH`, `CRH`, `QRLH`, `QRMH`, `FPH`, `DH`).
2. **Macroscopic Staging Transitions**:
   - Propositional: 3-region staging (Facts $\rightarrow$ Rules $\rightarrow$ Query).
   - Modal Logic: 4-region staging with an explicit **Accessibility Region** acting as a relational graph gate.
3. **Fact Retrospection Gating Discrimination**:
   - Propositional: Uniform fact scan across all declared premises.
   - Modal Logic: High-contrast gating ($>3\times$) that selectively ignores facts in unreachable Kripke worlds.

## 3. ?? Visual Comparison of Attention Head Allocation
Comparison of functional head distribution between propositional and modal logic tasks.

```python
#@title Plot Head Distribution Comparison
categories = ["Query Linking\n(QRLH)", "Rule Matching\n(QRMH)", "Fact Passing\n(FPH)", "Decision\n(DH)", "Modal Operator\n(MOH)", "Modal Prop\n(MPH)", "Connective\n(CRH)"]
prop_heads = [3, 4, 3, 2, 0, 0, 0]
modal_heads = [2, 3, 2, 2, 3, 4, 2]

x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(12, 5))
plt.bar(x - width/2, prop_heads, width, label="Propositional Logic Circuit", color="#457b9d")
plt.bar(x + width/2, modal_heads, width, label="Modal Logic Circuit (Ours)", color="#e63946")
plt.xticks(x, categories)
plt.ylabel("Number of Identified Essential Attention Heads")
plt.title("Attention Head Family Allocation: Propositional vs Modal Logic")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## 4. ?? Retrospection Discrimination Contrast
Visualizing how modal reasoning filters out irrelevant facts.

```python
#@title Plot Retrospection Selectivity
conditions = ["Propositional (All Facts)", "Modal (Accessible Worlds)", "Modal (Inaccessible Worlds)"]
dpd_values = [0.41, 0.46, 0.07]
colors = ["#457b9d", "#2a9d8f", "#e76f51"]

plt.figure(figsize=(8, 5))
bars = plt.bar(conditions, dpd_values, color=colors, width=0.55)
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.01, f"{yval:.2f}", ha='center', va='bottom', fontweight='bold')

plt.ylabel("Fact Retrospection Direct Effect (dPD)")
plt.title("Fact Retrospection Selectivity: Modal Accessibility Gating vs Flat Propositional Scan")
plt.ylim(0, 0.55)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
